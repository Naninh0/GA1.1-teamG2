```python
import numpy as np
```

```python
def substract(a, b):
    return a - b
```

```python
print(substract(5, 3))
print(substract(67, 2))
print(substract(100, 50))
```

```python
def distance_from_zero(x):
    if x < 0:
        return substract(0, x)
    else:
        return x
```

```python
distance_from_zero(4.8)
```

```python
distance_from_zero(-6.8)
```
